```python
from random import randint
```

```python
a_p = 0
b_p = 0
d = 0

while True:
    fordulok = int(input("Fordulók száma: "))
    if fordulok > 0:
        break
    print("Legalább 1 fordulót kell megadni!")

for i in range(fordulok):
    a_t = randint(1, 6)
    b_t = randint(1, 6)
    print(f"{i + 1}. forduló: Anna: {a_t}; Béla: {b_t} - ", end="")
    if a_t > b_t:
        a_p += 1
        print("Anna nyert")
    elif a_t < b_t:
        b_p += 1
        print("Béla nyert")
    else:
        d += 1
        print("Döntetlen")

print(f"Anna: {a_p} pont; Béla: {b_p} pont; Döntetlenek: {d}")
if a_p > b_p:
    print("A játék győztese: Anna")
elif b_p > a_p:
    print("A játék győztese: Béla")
else:
    print("A játék döntetlen.")
```

```python
print("Gondoltam egy számra 1 és 100 között.")
gondolt_szam = randint(1, 100)
probalkozas = 0

while True:
    tipp = int(input("Tipped: "))
    probalkozas += 1

    if tipp < gondolt_szam:
        print("Nagyobbra gondoltam!")
    elif tipp > gondolt_szam:
        print("Kisebbre gondoltam!")
    else:
        print("Gratulálok! Eltaláltad!")
        break

print(f"Próbálkozások száma: {probalkozas}")
```

```python
print("KŐ-PAPÍR-OLLÓ BAJNOKSÁG")
fordulok = int(input("Fordulók száma: "))
jatekos_pont = 0
gep_pont = 0
dontetlen = 0
nevek = ["kő", "papír", "olló"]

for i in range(fordulok):
    while True:
        valasztas = int(input(f"{i + 1}. forduló Választásod (1-kő, 2-papír, 3-olló): "))
        if 1 <= valasztas <= 3:
            break
        print("1, 2 vagy 3 közül válassz!")

    gep = randint(1, 3)
    print(f"Gép választása: {gep} ({nevek[gep - 1]}) - ", end="")

    if valasztas == gep:
        dontetlen += 1
        print("Döntetlen!")
    elif (valasztas == 1 and gep == 3) or (valasztas == 2 and gep == 1) or (valasztas == 3 and gep == 2):
        jatekos_pont += 1
        print("Nyertél!")
    else:
        gep_pont += 1
        print("A gép nyert!")

print(f"Végeredmény: Játékos {jatekos_pont}, Gép {gep_pont}, Döntetlen {dontetlen}")
```

```python
print("HÉT SZERENCSÉS")
dobasok = int(input("Dobások száma: "))
osszes_pont = 0
hetesek = 0
legjobb = 0

for i in range(dobasok):
    elso = randint(1, 6)
    masodik = randint(1, 6)
    osszeg = elso + masodik

    if osszeg == 7:
        pont = 10
        hetesek += 1
    elif elso == masodik:
        pont = 5
    else:
        pont = 1

    osszes_pont += pont
    if pont > legjobb:
        legjobb = pont

    print(f"{i + 1}. dobás: {elso} + {masodik} = {osszeg} -> {pont} pont")

print(f"Összpontszám: {osszes_pont}; Hetes összegek: {hetesek}; Legjobb dobás: {legjobb} pont")
```

```python
print("SZERENCSEKERÉK")
porgetesek = int(input("Pörgetések száma: "))
pontszam = 500
nullazasok = 0

for i in range(porgetesek):
    mezo = randint(1, 10)

    if 1 <= mezo <= 3:
        pontszam += 100
        eredmeny = "+100"
    elif 4 <= mezo <= 6:
        pontszam += 200
        eredmeny = "+200"
    elif 7 <= mezo <= 8:
        pontszam -= 150
        eredmeny = "-150"
    elif mezo == 9:
        pontszam *= 2
        eredmeny = "duplázás"
    else:
        pontszam = 0
        nullazasok += 1
        eredmeny = "nullázás"

    print(f"{i + 1}. pörgetés: {mezo} -> {eredmeny}, egyenleg: {pontszam}")

print(f"Végső pontszám: {pontszam}; Nullázások száma: {nullazasok}")
```

```python
print("PÉNZFELDOBÓ BAJNOKSÁG")
egyenleg = 1000
nyert_korok = 0
folytat = 1

while egyenleg > 0 and folytat == 1:
    print(f"Egyenleg: {egyenleg} Ft")

    while True:
        tipp = int(input("Fej (1) vagy írás (2)? "))
        if tipp == 1 or tipp == 2:
            break
        print("1-et vagy 2-t adj meg!")

    while True:
        tet = int(input("Tét: "))
        if 0 < tet <= egyenleg:
            break
        print("A tét legyen pozitív, és ne legyen nagyobb az egyenlegnél!")

    dobas = randint(1, 2)
    if dobas == 1:
        print("Dobás: fej", end=" ")
    else:
        print("Dobás: írás", end=" ")

    if tipp == dobas:
        egyenleg += tet
        nyert_korok += 1
        print(f"Nyertél {tet} Ft-ot!")
    else:
        egyenleg -= tet
        print(f"Vesztettél {tet} Ft-ot!")

    if egyenleg > 0:
        while True:
            folytat = int(input("Folytatod? (1-igen, 0-nem): "))
            if folytat == 0 or folytat == 1:
                break
            print("1-et vagy 0-t adj meg!")

print(f"Végső egyenleg: {egyenleg} Ft; Megnyert körök: {nyert_korok}")
```

```python
print("HUSZONEGY")
osszpont = 0
dobasok = 0

while osszpont < 21:
    while True:
        kerdes = int(input("Szeretnél dobni? (1-igen, 0-nem): "))
        if kerdes == 0 or kerdes == 1:
            break
        print("1-et vagy 0-t adj meg!")

    if kerdes == 0:
        break

    elso = randint(1, 6)
    masodik = randint(1, 6)
    osszeg = elso + masodik
    osszpont += osszeg
    dobasok += 1
    print(f"Dobás: {elso} + {masodik} = {osszeg}; Összpontszám: {osszpont}")

    if osszpont > 21:
        print("Túllépted a 21-et, vesztettél!")
    elif osszpont == 21:
        print("Pontosan 21 pont, nyertél!")

if osszpont < 21:
    print(f"Megálltál {osszpont} pontnál. {21 - osszpont} ponttal maradtál el a 21-től.")

print(f"Dobások száma: {dobasok}")
```

```python
print("KINCSVADÁSZAT")
print("Találd meg a kincset! 7 próbálkozásod van.")
kincs_x = randint(1, 10)
kincs_y = randint(1, 10)
probalkozasok = 0
megtalalta = False

while probalkozasok < 7 and not megtalalta:
    print(f"{probalkozasok + 1}. próbálkozás")

    while True:
        x = int(input("X koordináta (1-10): "))
        if 1 <= x <= 10:
            break
        print("A koordináta 1 és 10 közötti legyen!")

    while True:
        y = int(input("Y koordináta (1-10): "))
        if 1 <= y <= 10:
            break
        print("A koordináta 1 és 10 közötti legyen!")

    probalkozasok += 1

    if x == kincs_x and y == kincs_y:
        print(f"Gratulálok! Megtaláltad a kincset! Próbálkozások száma: {probalkozasok}")
        megtalalta = True
    else:
        tavolsag = ((kincs_x - x) ** 2 + (kincs_y - y) ** 2) ** 0.5
        print(f"Távolság: {tavolsag:.2f}", end=" ")

        if tavolsag <= 2:
            print("Forró!")
        elif tavolsag <= 5:
            print("Langyos!")
        else:
            print("Hideg!")

if not megtalalta:
    print(f"Nem találtad meg. A kincs helye: ({kincs_x}, {kincs_y})")
    print(f"Próbálkozások száma: {probalkozasok}")
```
